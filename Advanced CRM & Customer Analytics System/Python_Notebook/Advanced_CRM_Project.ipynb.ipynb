Handle Missing Values

```python
#We should not blindly delete missing records because that can cause loss of useful business information.
#Customers Dataset- Our customer dataset contains missing email addresses.
#We will replace missing emails with Not Available.
customers_clean['Email'] = (
    customers_clean['Email'].fillna('Not Available')
)

print("Missing customer emails handled!")
```

```python
#Transactions Dataset -The Transactions dataset contains missing payment methods.
transactions_clean['Payment_Method'] = (
    transactions_clean['Payment_Method'].fillna('Unknown')
)
```

```python
#Interactions Dataset-The Interactions dataset contains missing satisfaction ratings.
#We will use the median rating.
median_rating = interactions_clean[
    'Satisfaction_Rating'
].median()

interactions_clean[
    'Satisfaction_Rating'
] = interactions_clean[
    'Satisfaction_Rating'
].fillna(median_rating)

print("Missing satisfaction ratings handled!")
```

```python
#Support Tickets Dataset-Our Support Tickets dataset contains missing priority values.
#We will replace them with the most frequently occurring priority.
most_common_priority = support_tickets_clean[
    'Priority'
].mode()[0]

support_tickets_clean[
    'Priority'
] = support_tickets_clean[
    'Priority'
].fillna(most_common_priority)

print("Missing ticket priorities handled!")
```

```python
#Products Dataset-Our Products dataset doesn't intentionally contain missing values.
#we should check:
products_clean.isnull().sum()
```

Detect Outliers Using NumPy and Pandas

```python
"""What are outliers?

Outliers are values that are unusually high or unusually low compared with the majority of observations.

For example:

Suppose transaction amounts are:

₹500, ₹700, ₹900, ₹1,000, ₹1,200, ₹50,000.

₹50,000 might be an outlier.

However, an outlier is not automatically an error. It could be a genuine high-value transaction.
We will use the IQR method.
IQR = Q3 − Q1

Where:
- Q1 = 25th percentile.
- Q3 = 75th percentile."""
```

```python
#Calculate Q1 and Q3
Q1 = transactions_clean[
    'Transaction_Amount'
].quantile(0.25)

Q3 = transactions_clean[
    'Transaction_Amount'
].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)

print("Q3:", Q3)

print("IQR:", IQR)
```

Calculate lower and upper limits

```python
lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)

print("Upper Limit:", upper_limit)
```

Identify outliers

```python
outliers = transactions_clean[
    (transactions_clean['Transaction_Amount'] < lower_limit) |
    (transactions_clean['Transaction_Amount'] > upper_limit)
]

print("Total Outliers:", len(outliers))

outliers.head()
```

**Customer Segmentation & RFM Analysis**

```python
"""In this phase, we will build an advanced Customer Segmentation Model for your CRM Analytics project using Python, Pandas and NumPy.
This is one of the most important parts of your project because businesses need to understand which customers are valuable, which customers purchase frequently, and which customers are at risk of leaving."""
```

```python
#Prepare Transaction Data
"""We need completed transactions because we want to calculate actual purchasing behaviour"""
rfm_data = transactions_clean[
    transactions_clean['Order_Status'] == 'Completed'
].copy()

print("Total Completed Transactions:", len(rfm_data))

rfm_data.head()
```

```python
# Check Date Format
rfm_data['Transaction_Date'] = pd.to_datetime(
    rfm_data['Transaction_Date']
)

print(rfm_data['Transaction_Date'].dtype)
```

```python
#Define the Analysis Date
"""Recency measures how many days have passed since the customer's last purchase.
To calculate this, we need an analysis date.
We will use one day after the latest transaction date in the dataset."""

analysis_date = rfm_data['Transaction_Date'].max() + pd.Timedelta(days=1)

print("Analysis Date:", analysis_date)
```

```python
#Calculate RFM Values
"""We will calculate three values for every customer."""
rfm = rfm_data.groupby('Customer_ID').agg({

    'Transaction_Date': lambda x: (analysis_date - x.max()).days,

    'Transaction_ID': 'nunique',

    'Transaction_Amount': 'sum'

}).reset_index()
```

```python
#Rename the column
"""Interpretation:
- Recency 15 means the customer last purchased 15 days ago.
- Frequency 5 means the customer completed 5 orders.
- Monetary 12500 means the customer spent ₹12,500.
Each customer now has an RFM profile."""

rfm.columns = [
    'Customer_ID',
    'Recency',
    'Frequency',
    'Monetary'
]

rfm.head(10)
```

```python
#Understand RFM Statistics
"""Before creating customer segments, understand the distribution of your RFM values."""
rfm.describe()

```

```python
#Recency Distribution
plt.figure(figsize=(8,5))

sns.histplot(rfm['Recency'], bins=30, kde=True)

plt.title('Customer Recency Distribution')
plt.xlabel('Days Since Last Purchase')
plt.ylabel('Number of Customers')

plt.show()

#Frequency Distribution
plt.figure(figsize=(8,5))

sns.histplot(rfm['Frequency'], bins=30, kde=True)

plt.title('Customer Purchase Frequency')
plt.xlabel('Number of Purchases')
plt.ylabel('Number of Customers')

plt.show()

#Monetary Distribution
plt.figure(figsize=(8,5))

sns.histplot(rfm['Monetary'], bins=30, kde=True)

plt.title('Customer Monetary Distribution')
plt.xlabel('Total Customer Spending')
plt.ylabel('Number of Customers')

plt.show()
```

Create RFM Scores

```python
#Recency Score- For Recency, lower values are better.
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)
```

```python
#Frequency Score- For Frequency, higher values are better.
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

```python
#Monetary Score- For Monetary, higher values are better.
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

```python
#Check the results
rfm.head(10)
```

Calculate the Overall RFM Score

```python
#Combine all three scores
rfm['RFM_Total_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)
```

```python
#We will also create a combined RFM code.
rfm['RFM_Segment_Code'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)

rfm.head(10)
```

Create Customer Segments

```python
#Now we will assign meaningful business categories to customers.
def customer_segment(row):

    if row['R_Score'] >= 4 and row['F_Score'] >= 4 and row['M_Score'] >= 4:
        return 'Champions'

    elif row['F_Score'] >= 4 and row['R_Score'] >= 2:
        return 'Loyal Customers'

    elif row['R_Score'] >= 4 and row['F_Score'] <= 2:
        return 'New Customers'

    elif row['R_Score'] >= 3 and row['F_Score'] >= 2:
        return 'Potential Loyalists'

    elif row['R_Score'] <= 2 and row['F_Score'] >= 3:
        return 'At Risk'

    elif row['R_Score'] <= 2 and row['F_Score'] <= 2:
        return 'Hibernating'

    else:
        return 'Regular Customers'
```

```python
"""What have we achieved?
We have converted numerical RFM scores into meaningful customer categories.
For example, instead of simply knowing that a customer has an RFM score of 555, the business can identify that customer as a Champion.
These are simplified, custom segmentation rules for this project, not universal industry definitions."""

rfm['Customer_Segment'] = rfm.apply(
    customer_segment,
    axis=1
)

rfm.head(10)

```

Prepare the Transaction Dataset

```python
#Prepare completed transactions- We will analyse completed purchases only.
retention_data = transactions_clean[
    transactions_clean['Order_Status'] == 'Completed'
].copy()

retention_data['Transaction_Date'] = pd.to_datetime(
    retention_data['Transaction_Date']
)

print("Total Completed Transactions:", len(retention_data))

retention_data.head()
```

One-Time vs Repeat Customers

```python
#Count purchases by customer
customer_purchase_count = (
    retention_data
    .groupby('Customer_ID')['Transaction_ID']
    .nunique()
    .reset_index()
)

customer_purchase_count.columns = [
    'Customer_ID',
    'Total_Purchases'
]

customer_purchase_count.head()
```

```python
#Create customer purchase categories
def purchase_type(purchases):

    if purchases == 1:
        return 'One-Time Customer'

    else:
        return 'Repeat Customer'

#Apply Function
customer_purchase_count['Purchase_Type'] = (
    customer_purchase_count['Total_Purchases']
    .apply(purchase_type)
)

customer_purchase_count.head(10)
```

```python
#Count one-time and repeat customers
purchase_type_count = customer_purchase_count[
    'Purchase_Type'
].value_counts()

print(purchase_type_count)

#Visualise one-time vs repeat customers
plt.figure(figsize=(8,5))

sns.barplot(
    x=purchase_type_count.index,
    y=purchase_type_count.values
)

plt.title('One-Time vs Repeat Customers')
plt.xlabel('Customer Type')
plt.ylabel('Number of Customers')

plt.show()
```

```python
#Calculate repeat customer percentage
total_customers = len(customer_purchase_count)

repeat_customers = (
    customer_purchase_count['Purchase_Type'] == 'Repeat Customer'
).sum()

one_time_customers = (
    customer_purchase_count['Purchase_Type'] == 'One-Time Customer'
).sum()

repeat_rate = (repeat_customers / total_customers) * 100

one_time_rate = (one_time_customers / total_customers) * 100

print("Total Customers:", total_customers)

print("Repeat Customers:", repeat_customers)

print("One-Time Customers:", one_time_customers)

print("Repeat Customer Rate:", round(repeat_rate, 2), "%")

print("One-Time Customer Rate:", round(one_time_rate, 2), "%")
```

Customer Retention Cohort Analysis

```python
#Find each customer's first purchase month
retention_data['Transaction_Month'] = (
    retention_data['Transaction_Date'].dt.to_period('M')
)

customer_cohort = (
    retention_data
    .groupby('Customer_ID')['Transaction_Month']
    .min()
    .reset_index()
)

customer_cohort.columns = [
    'Customer_ID',
    'Cohort_Month'
]

customer_cohort.head()
```

```python
#Merge cohort information with transactions
cohort_data = retention_data.merge(
    customer_cohort,
    on='Customer_ID',
    how='left'
)

cohort_data.head()
```

```python
#Calculate cohort index
cohort_data['Cohort_Index'] = (
    (cohort_data['Transaction_Month'].dt.year -
     cohort_data['Cohort_Month'].dt.year) * 12
    +
    (cohort_data['Transaction_Month'].dt.month -
     cohort_data['Cohort_Month'].dt.month)
    + 1
)

cohort_data.head()
```

```python
#Calculate monthly cohort customers
cohort_table = cohort_data.pivot_table(
    index='Cohort_Month',
    columns='Cohort_Index',
    values='Customer_ID',
    aggfunc='nunique'
)

print(cohort_table)
```

```python
#Calculate cohort retention percentage- We divide each month's active customers by the original cohort size.
cohort_sizes = cohort_table.iloc[:, 0]

retention_matrix = cohort_table.divide(
    cohort_sizes,
    axis=0
) * 100

print(retention_matrix.round(2))
```

```python
#Create a cohort retention heatmap
plt.figure(figsize=(14,8))

sns.heatmap(
    retention_matrix,
    annot=True,
    fmt='.1f',
    cmap='YlGnBu',
    linewidths=0.5
)

plt.title('Monthly Customer Retention Cohort Analysis')
plt.xlabel('Months Since First Purchase')
plt.ylabel('Customer Acquisition Cohort')

plt.show()
```

Identify Customers at Churn Risk

```python

"""Customer churn occurs when customers stop purchasing from a business.
Our dataset does not contain a confirmed churn label. Therefore, we will create a churn-risk indicator based on purchase inactivity."""
```

```python
#Calculate the analysis date
analysis_date = (
    retention_data['Transaction_Date'].max()
    + pd.Timedelta(days=1)
)

print("Analysis Date:", analysis_date)
```

```python
#Calculate the last purchase date of every customer
customer_last_purchase = (
    retention_data
    .groupby('Customer_ID')['Transaction_Date']
    .max()
    .reset_index()
)

customer_last_purchase.columns = [
    'Customer_ID',
    'Last_Purchase_Date'
]

customer_last_purchase.head()
```

```python
#Calculate inactive days
customer_last_purchase['Days_Inactive'] = (
    analysis_date -
    customer_last_purchase['Last_Purchase_Date']
).dt.days

customer_last_purchase.head()
```

```python
#Create churn-risk categories
def churn_risk(days):

    if days <= 30:
        return 'Active'

    elif days <= 60:
        return 'Low Risk'

    elif days <= 90:
        return 'Medium Risk'

    else:
        return 'High Risk'

#Apply
customer_last_purchase['Churn_Risk'] = (
    customer_last_purchase['Days_Inactive']
    .apply(churn_risk)
)

customer_last_purchase.head(10)
```

```python
#Count customers by churn risk
churn_distribution = (
    customer_last_purchase['Churn_Risk']
    .value_counts()
)

print(churn_distribution)
```

```python
#Visualise churn risk
plt.figure(figsize=(9,5))

sns.countplot(
    data=customer_last_purchase,
    x='Churn_Risk',
    order=['Active', 'Low Risk', 'Medium Risk', 'High Risk']
)

plt.title('Customer Churn Risk Analysis')
plt.xlabel('Churn Risk')
plt.ylabel('Number of Customers')

plt.show()
```

Support Ticket Resolution Time Analysis

```python
#Create a resolved ticket dataset
resolved_tickets = ticket_analysis.dropna(
    subset=['Resolved_Date']
).copy()

print("Total Resolved Tickets:", len(resolved_tickets))
```

```python
#Calculate resolution time
# Formula:Resolution Time = Resolved Date − Created Date
resolved_tickets['Resolution_Days'] = (
    resolved_tickets['Resolved_Date'] -
    resolved_tickets['Created_Date']
).dt.total_seconds() / 86400

#Check Result
resolved_tickets[
    ['Ticket_ID', 'Created_Date', 'Resolved_Date', 'Resolution_Days']
].head(10)
```

```python
#Calculate average resolution time
average_resolution_time = (
    resolved_tickets['Resolution_Days'].mean()
)

print(
    "Average Ticket Resolution Time:",
    round(average_resolution_time, 2),
    "days"
)
```

```python
#Analyse resolution time by priority
resolution_by_priority = (
    resolved_tickets
    .groupby('Priority')['Resolution_Days']
    .agg(['mean', 'median', 'count'])
    .sort_values('mean', ascending=False)
)

print(resolution_by_priority)

#Visualise
plt.figure(figsize=(9,5))

sns.barplot(
    data=resolved_tickets,
    x='Priority',
    y='Resolution_Days',
    errorbar=None,
    order=['Critical','High', 'Medium', 'Low']
)

plt.title('Average Resolution Time by Ticket Priority')
plt.xlabel('Priority')
plt.ylabel('Resolution Time (Days)')

plt.show()
```

SLA Performance Analysis


```python
"""What is SLA?
SLA stands for Service Level Agreement.

It defines how quickly a company should respond to or resolve a customer issue."""
```

```python
#Create SLA target column
sla_targets = {
    'High': 1,
    'Medium': 3,
    'Low': 5
}

resolved_tickets['SLA_Target_Days'] = (
    resolved_tickets['Priority'].map(sla_targets)
)
```

```python
#Compare actual resolution time with SLA
resolved_tickets['SLA_Status'] = np.where(
    resolved_tickets['Resolution_Days'] <=
    resolved_tickets['SLA_Target_Days'],
    'Within SLA',
    'SLA Breached'
)

#Check
print(
    resolved_tickets[
        ['Ticket_ID', 'Priority', 'Resolution_Days',
         'SLA_Target_Days', 'SLA_Status']
    ].head(10)
)
```

```python
#Calculate SLA performance
sla_distribution = (
    resolved_tickets['SLA_Status']
    .value_counts()
)

print(sla_distribution)

#Calculate SLA compliance
sla_compliance = (
    (resolved_tickets['SLA_Status'] == 'Within SLA').mean()
) * 100

print("SLA Compliance:", round(sla_compliance, 2), "%")
```

```python
#Visualise SLA performance
plt.figure(figsize=(8,5))

sns.countplot(
    data=resolved_tickets,
    x='SLA_Status'
)

plt.title('SLA Performance Analysis')
plt.xlabel('SLA Status')
plt.ylabel('Number of Tickets')

plt.show()
```
